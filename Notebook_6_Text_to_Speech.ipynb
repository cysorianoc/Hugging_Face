### Build the `text-to-speech` pipeline using the 🤗 Transformers Library

- Here is some code that suppresses warning messages.

````python
from transformers.utils import logging

logging.set_verbosity_error()
````

````python
from transformers import pipeline

narrator = pipeline("text-to-speech",
                    model="kakao-enterprise/vits-ljs")
````

Info about [kakao-enterprise/vits-ljs](https://huggingface.co/kakao-enterprise/vits-ljs)

````python
text = """
Researchers at the Allen Institute for AI, \
HuggingFace, Microsoft, the University of Washington, \
Carnegie Mellon University, and the Hebrew University of \
Jerusalem developed a tool that measures atmospheric \
carbon emitted by cloud servers while training machine \
learning models. After a model’s size, the biggest variables \
were the server’s location and time of day it was active.
"""
````

````python
narrated_text = narrator(text)
````

````python
from IPython.display import Audio as IPythonAudio

IPythonAudio(narrated_text["audio"],
             rate=narrated_text["sampling_rate"])
````

There are two variants of the VITS model: one is trained on the LJ Speech dataset, and the other is trained on the VCTK dataset. LJ Speech dataset consists of 13,100 short audio clips of a single speaker with a total length of approximately 24 hours. The VCTK dataset consists of approximately 44,000 short audio clips uttered by 109 native English speakers with various accents. The total length of the audio clips is approximately 44 hours.

Let's see the result with ```vits-vctk``` variant

````python
from transformers import pipeline

pipe = pipeline("text-to-speech", model="kakao-enterprise/vits-vctk")
````

````python
narrated_text=pipe(text)
````

````python
from IPython.display import Audio as IPythonAudio

IPythonAudio(narrated_text["audio"],
             rate=narrated_text["sampling_rate"])
````

Finally let's try ```wfloat``` model.

wfloat-tts is a lightweight multi-speaker English VITS text-to-speech model with speaker, emotion, and intensity control


[tSource](https://huggingface.co/Wfloat/wfloat-tts)

````python
!pip install wfloat
````

````python
import wfloat
from IPython.display import Audio as IPythonAudio

model = wfloat.load("wfloat/wfloat-tts")

result = model.generate(
    text="Hey there, how are you today?",
    voice_id="narrator_woman",
    emotion="neutral",
    intensity=0.5,
)

# Save to WAV
result.audio.save("out.wav")

# Play audio in the notebook
IPythonAudio(
    result.audio.samples,
    rate=result.audio.sample_rate
)
````

````python
text_test = """
Researchers at the Allen Institute for AI, \
HuggingFace, Microsoft, the University of Washington, \
Carnegie Mellon University, and the Hebrew University of \
Jerusalem developed a tool that measures atmospheric \
carbon emitted by cloud servers while training machine \
learning models. After a model’s size, the biggest variables \
were the server’s location and time of day it was active.
"""
````
